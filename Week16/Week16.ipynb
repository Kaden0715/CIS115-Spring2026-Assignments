```python
# NAME: expense_tracker.py

# IMPORT STATEMENTS
# (None)

# FUNCTION DEFINITIONS
# (None)

# PROGRAM LOGIC
# 1. Create empty list to store expenses
expenses = []
# 2. Use while loop to keep program running
while True:
# 3. Show menu options (Add, Get Totals, Exit)
    print("\n--- Expense Tracker ---")
    print("1. Add New Expense")
    print("2. Get Totals")
    print("3. Exit")
    choice = input("Selection: ")
# 4. Add new expense
    if choice == "1":
# 5. Ask for category, vendor, amount, date
        category = input("Enter Category: ")
        vendor = input("Enter Vendor: ")
        amount = float(input("Enter Amount: "))
        date = input("Enter Date: ")
# 6. Create dictionary for one expense
        expense = {
            "category": category,
            "vendor": vendor,
            "amount": amount,
            "date": date
        }
# 7. Add dictionary to list
        expenses.append(expense)
# 8. Get totals option
    elif choice == "2":
# 9. Show options (All, Category, Vendor)
        print("\n--- Totaling Options ---")
        print("A: All Expenses")
        print("C: Filter by Category")
        print("V: Filter by Vendor")
        option = input("Choice: ")
# 10. All expenses total
        if option == "A":
# 11. Loop through list and add all amounts
            total = 0
            for e in expenses:
                total = total + e["amount"]
# 18. Print total formatted to 2 decimals
            print("Total: $" + format(total, ".2f"))
# 12. Category total
        elif option == "C":
# 13. Ask for category
            cat = input("Enter Category to sum: ")
# 14. Loop and only add matching category
            total = 0
            for e in expenses:
                if e["category"] == cat:
                    total = total + e["amount"]
            print("The total for " + cat + " is: $" + format(total, ".2f"))
# 15. Vendor total
        elif option == "V":
# 16. Ask for vendor
            ven = input("Enter Vendor to sum: ")
# 17. Loop and only add matching vendor
            total = 0
            for e in expenses:
                if e["vendor"] == ven:
                    total = total + e["amount"]
# 18. Print total formatted to 2 decimals
            print("The total for " + ven + " is: $" + format(total, ".2f"))
# 19. Exit program
    elif choice == "3":
        break
```

```python
# NAME: study_tracker.py

# IMPORT STATEMENTS
# (None)

# PROGRAM LOGIC
sessions = []

while True:
    print("\n1. Add Study Session")
    print("2. View Sessions")
    print("3. Total Time")
    print("4. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        subject = input("Enter subject: ")

        try:
            minutes = int(input("Enter minutes studied: "))
        except:
            print("Invalid input. Please enter a number.")
            continue

        sessions.append((subject, minutes))

        file = open("study_log.txt", "a")
        file.write(subject + "," + str(minutes) + "\n")
        file.close()

    elif choice == "2":
        print("\n--- Study Sessions ---")
        for s in sessions:
            print(s[0] + ": " + str(s[1]) + " minutes")

    elif choice == "3":
        total = 0
        for s in sessions:
            total = total + s[1]
        print("Total study time: " + str(total) + " minutes")

    elif choice == "4":
        break

    else:
        print("Invalid choice. Try again.")
```

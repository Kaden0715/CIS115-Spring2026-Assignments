```python
# NAME: study_tracker.py

# IMPORT STATEMENTS
# (None)

# PROGRAM LOGIC
sessions = []

while True:
    print("\n1. Add Study Session")
    print("2. View Sessions")
    print("3. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        subject = input("Enter subject: ")
        minutes = int(input("Enter minutes studied: "))
        sessions.append((subject, minutes))

    elif choice == "2":
        print("\n--- Study Sessions ---")
        for s in sessions:
            print(s[0] + ": " + str(s[1]) + " minutes")

    elif choice == "3":
        break
```

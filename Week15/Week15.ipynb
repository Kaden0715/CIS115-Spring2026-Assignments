```python
# NAME: task_manager.py

# IMPORT STATEMENTS
# (None)

# PROGRAM LOGIC
# 1. Create empty task list
tasks = []
# 2. Use loop to keep program running
while True:
# 3. Ask user for action
    print("\n1. Add Task")
    print("2. View Tasks")
    print("3. Exit")
    choice = input("Choose an option: ")
# 4. Add task
    if choice == "1":
        name = input("Enter task name: ")
        priority = input("Enter priority (1=High, 2=Medium, 3=Low): ")
        tasks.append((name, priority))
# 5. View tasks
    elif choice == "2":
        print("\n--- Task List ---")
        for t in tasks:
            print(t[0] + " (Priority " + t[1] + ")")
# 6. Exit
    elif choice == "3":
        break
```

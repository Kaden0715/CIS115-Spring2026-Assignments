```python
# NAME: task_manager.py

# IMPORT STATEMENTS
# (None)

# PROGRAM LOGIC
# 1. Create empty task list
tasks = []
# 2. Use loop to keep program running
while True:
# 3. Ask user for action
    print("\n1. Add Task")
    print("2. View Tasks")
    print("3. Complete Task")
    print("4. Exit")
    choice = input("Choose an option: ")
# 4. Add task
    if choice == "1":
        name = input("Enter task name: ")
        priority = input("Enter priority (1=High, 2=Medium, 3=Low): ")
        tasks.append((name, priority))
# 5. View tasks
    elif choice == "2":
        print("\n--- Task List ---")
        for i in range(len(tasks)):
            print(str(i) + ". " + tasks[i][0] + " (Priority " + tasks[i][1] + ")")
# 6. Complete task
    elif choice == "3":
        index = int(input("Enter task number to complete: "))
        completed = tasks.pop(index)
# 7. Save completed task to file
        file = open("history.txt", "a")
        file.write(completed[0] + "\n")
        file.close()

        print("Task completed and saved.")
# 8. Exit
    elif choice == "4":
        break
```
